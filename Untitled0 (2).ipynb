```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for i in df.columns:
  if df[i].dtype == 'object':
    df[i] = le.fit_transform(df[i])
```

```python
df.head()
```

```python
x = df.drop('salary_in_usd',axis = 1)
y = df['salary_in_usd']
#
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
#
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
X_train = sc.fit_transform(x_train)
X_test = sc.transform(x_test)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.metrics import mean_squared_error,r2_score , root_mean_squared_error
def evaluate(model,x_train,x_test,y_train,y_test):
  train_pred = model.predict(x_train)
  test_pred = model.predict(x_test)

  # r2_scor
  print(f'train r2 score : {model} is  {r2_score(y_train,train_pred)}')
  print(f'test r2 score : {model} is{r2_score(y_test,test_pred)}')

  # mean_squared_error
  print(f'train mse score error : {model} is  {mean_squared_error(y_train,train_pred)}')
  print(f'test mse score error : {model} is   {mean_squared_error(y_test,test_pred)}')

  # root_mean_squared_error
  print(f'root_mean_squared_error : {model} is {root_mean_squared_error(y_train,train_pred)}')
  print(f'root_mean_squared_error : {model} is {root_mean_squared_error(y_test,test_pred)}')
```

```python
from sklearn.linear_model import LinearRegression,Ridge
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
evaluate(lr,X_train,X_test,y_train,y_test)
```

```python
y_pred = lr.predict(X_test)
```

```python
y_pred_train = lr.predict(X_train)
```

```python
from sklearn.metrics import r2_score
r2_score(y_train,y_pred_train)  #Testing data accuracy (-)
r2_score(y_test,y_pred) #Training data accuracy(+)
```
